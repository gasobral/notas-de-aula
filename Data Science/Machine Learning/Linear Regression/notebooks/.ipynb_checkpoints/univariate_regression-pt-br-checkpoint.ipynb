# Regressão Linear
A *regressão linear* é um modelo bastante utilizado pois ele é intuitivo
e sua interpretação não é complicada. Por exemplo, temos a intuição de
que o gasto de uma pessoa tem uma relação com a sua renda. Para
quantificarmos essa relação, podemos utilizar uma regressão a fim de
expressar o gasto em função da renda. Podemos escrever isso como

$$ gasto = renda \times \beta$$

onde $\beta$ é um valor real. Da forma como escrevemos a conta, estamos
assumindo que exista uma proporção, expresso pelo $\beta$, entre gasto e
renda, em outras palavras, que existe uma relação linear entre renda e
gasto. Esse fato é bastante importante na regressão linear.

Agora, para estabelecermos um modelo a partir desta conta, vamos
adicionar um valor real $\alpha$, nesse contexto ele representa o gasto
quando a renda é zero, e um valor real $e$ para representar o erro do
modelo. Desta forma, temos o seguinte:

$$ gasto = \alpha + renda \times \beta + e$$

Agora o nosso modelo está pronto. Note que podemos usá-lo para prever o
gasto a partir da renda. Desde que exista uma forte relação linear
(expresso pelo coeficiente de correlação) entre as variáveis e o erro
não seja grande. Algumas observações sobre o modelo:

- o gasto é a variável alvo (o que desejo prever);
- a renda é a variável preditora (variável usada para prever valores);
- o $\alpha$ e o $\beta$ são os parâmetros do modelo.

Você também pode encontrar a nomenclatura variável explicadora para
variável preditora. Mais adiante falaremos sobre os parâmetros do
modelo. No momento basta saber que, quando um modelo é treinado, são
determinados parâmetros que minimizem o erro $e$. Após vermos esse
exemplo, não é tão assustador definir a regressão linear como:

$$ y = \alpha + \beta x + e$$

onde $y$ é variável alvo e $x$ a variável preditora. O $\alpha$ é chamado
de intercepto, pois é o ponto em que reta cruza o eixo y. Adiante apresentarei
uma situação problema para abordamos a regressão linear na prática. Por
enquanto estamos lidando apenas com uma variável, mas abordaremos uma
generalização desse modelo. Usarei como referência os livros
[1], [2] e [3] para apresentar este modelo de forma exploratória.

[1] BUSSAB, W. O.; MORETTIN, P. A. Estatística Básica. Atual Editora, São Paulo, 1988

[2] GÉRON, Aurélien. Hands-On Machine Learning with Scikit-Lean, Keras and TensorFlow. O' Reilly Media Inc. 2021

[3] MORETTIN, Pedro Alberto; SINGER, Júlio da Motta. Estatística e ciência de dados. Rio de Janeiro, LTC, 2022

## Situação-problema
Uma empresa da área da saúde deseja monitorar a função respiratória de
funcionários expostos à poeira. Ela forneceu uma planília com os
seguintes dados:

| Nome da coluna | Descrição                           |
| :---           | :---                                |
| PEFR           | taxa do volume de ar expelido dos   |
|                | pulmões em uma expiração rápida,    |
|                | esse é um indicador confiável da    |
|                | ventilação e também do fluxo de ar. |
| Exposure       | tempo de exposição em anos à poeira.|

Deseja-se estimar o PEFR dado o tempo de exposição (Exposure).

A primeira tarefa a ser feita é entendermos os dados que estamos
lidando. O Exposure é bastante direto e não precisa de muitas

reais. Uma vez identificado o domínio dos dados, vamos examiná-los
usando a análise exploratória (hora de usarmos os conhecimentos em
estatística).

## Regressão Linear Simples (Univariada)
A regressão linear simples, ou univariada, é um modelo de regressão
definido com apenas uma variável preditora. No exemplo da empresa da
área da saúde, o tempo de exposição é a variável preditora. Sendo assim,
a equação abaixo define uma regressão linear simples para prever o valor
do PEFR:

$$ PEFR = \alpha + Exposure \times \beta + e$$

Agora veremos como podemos implementar esse modelo usando Python.

```python
from sklearn.linear_model import LinearRegression
```

```python
## vamos denotar por X as variáveis preditoras
X = lung_disease.drop(columns=['PEFR'])

## vamos denotar por y a variável alvo
y = lung_disease['PEFR']
```

```python
linear_model = LinearRegression()

## o método fit faz o treinamento do modelo
## isso significa que são otidos os parâmetros do modelo
linear_model.fit(X, y)
```

```python
## irei plotar a reta que represnta o modelo apenas para
## ter uma visualização do modelo
fig, ax = plt.subplots(1, 1, figsize=(14, 8))

ax.set_xlim(0, 23)
ax.set_ylim(0, 610)

ax.set_xlabel('Exposure')
ax.set_ylabel('PEFR')

## obtém os valores extremos da equação de regressão para desenhar os
## extremos da reta
ax.plot((0, 23),
        linear_model.predict(pd.DataFrame({'Exposure': [0, 23]})))

## posiciona o intercepto (o alfa da equação) na reta
ax.text(0.4,
        linear_model.intercept_,
        r'$\alpha$',
        size='larger')

plt.show()
```

```python
print(f'Equação: Y = {linear_model.intercept_:.3f} + '
      f'{linear_model.coef_[0]:.3f} X')
```

Após treinarmos o modelo, é interessante saber se as previsões feitas
por ele são precisas ou não. Uma abordagem bem direta seria prever
valores e compará-los com os dados de PEFR, que já temos. O trecho de
código a seguir faz essa comparação.

```python
## usamos o método predict para predizer os valores
## aqui basta usar valore que representem o tempo de
## exposição a poeira em anos
predicted_value = linear_model.predict(X)

## fazemos a conta abaixo para termos uma ideia
## do quanto o valor previsto está distante do valor real
error = y - predicted_value

for x, real, p, erro in zip(lung_disease.Exposure,
                            lung_disease.PEFR,
                            predicted_value,
                            error):
    print(f"X: {x}  Real value: {real}  Predicted value: {p}  Error: {erro}")
```

```python
## Esse trecho de código serve para visualizarmos o modelo e os
## seus erros de forma mais amigável
fig, ax = plt.subplots(1, 1, figsize=(14, 8))

## primeiro, simplesmente plotamos os pontos do gráfico de
## dispersão entre tempo de exposição (eixo x) x PEFR (eixo y)
ax.scatter(x=lung_disease['Exposure'],
           y=lung_disease['PEFR'])

## plota o segmento de reta que representa o modelo
## desenha cada um dos pontos (tempo de exposição, valor previsto)
ax.plot(lung_disease.Exposure, predicted_value)

## desejamos plotar, para cada valor previsto, uma reta tracejada em
## laranja representa a distância entre o valor real e o valor previsto
for x, y_real, y_predicted in zip(lung_disease.Exposure,
                                  lung_disease.PEFR,
                                  predicted_value):
    ## desenha um segmento de reta entre os seguintes pontos
    ## (tempo de exposição, PEFR real) -- (tempo de exposição, PEFR previsto)
    ax.plot((x,x), (y_real, y_predicted), '--', color='C1')

ax.set_title('Valor previsto x Valor real')
ax.set_xlabel('Exposição em anos')
ax.set_ylabel('PEFR')

plt.show()
```

A partir do gráfico acima, podemos ver que há alguns pontos em que nosso
modelo comete erros bem grosseiros. Por exemplo, com tempo de exposição
= 4 e PEFR = 100, o modelo apresenta um erro de $-297,844...$. Há outros
exemplos do gênero para tempo de exposição = 20. Isso nos sugere que há
casos extremos e deveríamos pensar em como lidar com eles. A primeira
sugestão seria eliminar os outliers, pois eles podem deslocar a reta
(que representa o modelo) e prejudicar a performance do modelo. Mas
o ponto crucial foi termos treinado o modelo em todos os dados
disponíveis.
